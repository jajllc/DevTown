```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel("DAY4.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Impute numerical null values with mean
df['sales']=df['sales'].fillna(df['sales'].mean())
```

```python
# Impute categorical values with amythimgg you want
df['product_name']=df['product_name'].fillna('unknown')
```

```python
df.duplicated().sum()
```

```python
# Remove duplicate records
df.drop_duplicates(inplace=True)
```

```python
# Rename
df.rename(columns={'product_name':'product'},inplace=True)
```

```python
df.columns
```

```python
# Changing data type
df['quantity']=df['quantity'].astype(int)
```

```python
# Pivot tables using groupby()
df.groupby("region")["sales"].sum().sort_values(ascending=False)
```

```python
# Profit by category
df.groupby("category")["profit"].sum().sort_values(ascending=False)
```

```python
# Top 10 products
df.groupby("product")["sales"].sum().sort_values(ascending=False).head(10)
```

```python
# visulaization
# sales by region
df.groupby("region")["sales"].sum().plot(kind="bar")
plt.title("Sales by region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
# Profit by category
df['profit'].groupby(df['category']).sum().plot(kind='bar')
plt.title("Profit by category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()

```

```python
df.groupby("category")["quantity"].sum().plot(kind="pie")
```
